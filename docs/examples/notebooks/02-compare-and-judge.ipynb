# Compare agents and add a judge

This notebook compares Codex and Claude Code on the same prompt, asks Claude to judge the labeled outputs, then opens the generated compare report.

```bash
%%bash
set -euo pipefail
mkdir -p /tmp/harnessctl-notebook-compare
cd /tmp/harnessctl-notebook-compare
git init >/dev/null 2>&1 || true
cat > README.md <<'EOF'
# Compare notebook fixture

This repo is intentionally small. Use it to compare agent behavior without touching a real project.
EOF
git add README.md
git commit -m 'Initial fixture' >/dev/null 2>&1 || true
pwd
```

Run the same review prompt across both harnesses and ask Claude Code to judge the results.

```bash
%%bash
cd /tmp/harnessctl-notebook-compare
harnessctl compare \
  "Review README.md. Do not edit files. Give one strength, one risk, and one concrete improvement." \
  --agents codex,claude \
  --judge claude
```

Find and inspect the generated report. It includes the prompt, labeled outputs, run IDs, and the judge verdict.

```bash
%%bash
cd /tmp/harnessctl-notebook-compare
REPORT=$(find .harnessctl/compare -type f | sort | tail -1)
echo "$REPORT"
sed -n '1,220p' "$REPORT"
```

Continue from the best result by using one of the run IDs in the report.

```bash
%%bash
cd /tmp/harnessctl-notebook-compare
harnessctl handoff RUN_ID_FROM_REPORT --agent claude \
  "Use the previous result as context and propose a final README improvement. Do not edit files."
```
