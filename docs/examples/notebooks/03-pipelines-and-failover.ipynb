# Pipelines and fallback basics

This notebook shows two common workflows: a sequential pipeline and a fallback chain. Use it to understand how harnessctl routes work between harnesses.

```bash
%%bash
set -euo pipefail
mkdir -p /tmp/harnessctl-notebook-pipeline
cd /tmp/harnessctl-notebook-pipeline
git init >/dev/null 2>&1 || true
cat > README.md <<'EOF'
# Pipeline notebook fixture

A tiny repo for testing plan/build/review agent chains.
EOF
git add README.md
git commit -m 'Initial fixture' >/dev/null 2>&1 || true
pwd
```

Run a two-stage pipeline. Codex plans, Claude Code reviews the plan, and both runs share one harness session.

```bash
%%bash
cd /tmp/harnessctl-notebook-pipeline
harnessctl pipeline \
  "Improve README.md for a user who is new to this repo. Do not edit files." \
  --step codex:"Propose a concise plan." \
  --step claude:"Review the plan and name the top risk." \
  --name notebook-pipeline
```

Configure a fallback chain. This example keeps it simple: use Codex by default and offer Claude Code as fallback.

```bash
%%bash
harnessctl config set default_agent codex
harnessctl config set-fallback codex claude
harnessctl config get-fallback codex
```

Check the configured agents before running a real task.

```bash
%%bash
harnessctl doctor
harnessctl list
```

Clean up the fallback if this was only a local experiment.

```bash
%%bash
harnessctl config remove-fallback codex
```
