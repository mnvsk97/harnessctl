# Quickstart: Codex to Claude Code handoff

This notebook walks through a small end-to-end handoff. It starts a task with Codex, passes the run to Claude Code, then checks the run chain.

Use a disposable git repo for this walkthrough. Codex expects a trusted git workspace.

```bash
%%bash
set -euo pipefail
mkdir -p /tmp/harnessctl-notebook-handoff
cd /tmp/harnessctl-notebook-handoff
git init >/dev/null 2>&1 || true
cat > README.md <<'EOF'
# Handoff notebook fixture

A tiny repo for testing harnessctl handoffs.
EOF
git add README.md
git commit -m 'Initial fixture' >/dev/null 2>&1 || true
pwd
```

Start with Codex. The command prints a run ID at the end; copy that value into the next cell.

```bash
%%bash
cd /tmp/harnessctl-notebook-handoff
harnessctl run --agent codex --name notebook-handoff \
  "Create notes.md with one heading named Step 1 - Codex and two bullets about what you changed."
```

Replace `RUN_ID_FROM_CODEX` with the run ID printed by the previous command.

```bash
%%bash
cd /tmp/harnessctl-notebook-handoff
harnessctl handoff RUN_ID_FROM_CODEX --agent claude --name notebook-handoff \
  "Read notes.md. Append a heading named Step 2 - Claude with two bullets. Keep it concise."
```

Inspect the resulting file and the linked run chain.

```bash
%%bash
cd /tmp/harnessctl-notebook-handoff
sed -n '1,120p' notes.md
harnessctl logs --limit 5
```
